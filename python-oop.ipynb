#### Section 1: Classes and Objects

```python
class Car:
    # A class attribute, shared by all instances of the class
    wheels = 4

    # The __init__ method is the constructor. It's called when you create a new object.
    # 'self' refers to the instance being created.
    def __init__(self, make, model, year):
        #instance attributes
        self.make = make
        self.model = model
        self.year = year

    # instance method
    def display_info(self):
        return(f"{self.year} {self.make} {self.model}")

# create instances of Car class
car1 = Car("VW", "Taos", "2023")
car2 = Car("Cadillac", "Lyriq", "2025")
```

```python
print(car1.display_info())
print(car2.display_info())
```
