#### 4

Given 2 strings, `a` and `b`, return the number of the positions where they contain the same length 2 substring. So "xxcaazz" and "xxbaaz" yields 3, since the "xx", "aa", and "az" substrings appear in the same place in both strings.


string_match('xxcaazz', 'xxbaaz') → 3

string_match('abc', 'abc') → 2

string_match('abc', 'axc') → 0

````python
def string_match(a, b):

    shorter = min(len(a), len(b))

    cnt = 0

    for i in range(0, shorter-1):
        a_sub = a[i:i+2]
        b_sub = b[i:i+2]
        
        if a_sub == b_sub:
          cnt = cnt + 1

    return cnt
````

#### 7 Centered Average

Return the "centered" average of an array of ints, which we'll say is the mean average of the values, except ignoring the largest and smallest values in the array. If there are multiple copies of the smallest value, ignore just one copy, and likewise for the largest value. Use int division to produce the final average. You may assume that the array is length 3 or more.


`centered_average([1, 2, 3, 4, 100]) → 3`

`centered_average([1, 1, 5, 5, 10, 8, 7]) → 5`

`centered_average([-10, -4, -2, -4, -2, 0]) → -3`

````python
def centered_average(nums):

    nums = sorted(nums)

    # find the min and max
    sm = min(nums)
    lg = max(nums)

    sm_idx = nums.index(sm)
    lg_idx = nums.index(lg)

    print("indices", sm_idx, lg_idx)

    if sm_idx != lg_idx:
        nums.pop(sm_idx)
        nums.pop(lg_idx-1) # since 0-indexed
    else:
        nums.pop(sm_idx)

    print(sum(nums)//len(nums))
````

#### 9 Sort Array by Parity

Given an integer array nums, move all the even integers at the beginning of the array followed by all the odd integers.

Return any array that satisfies this condition.

````python
def sortArrayByParity(nums):

    # even element % 2 == 0
    evens = [e for e in nums if e%2 == 0]
    # odd element % 2 != 0
    odds = [o for o in nums if o%2 != 0]

    return evens + odds
````

#### 11 Binary Search `O(\log n)`time complexity

1. Initialize Pointers → left = 0, right = len(nums) - 1
2.	Find the Middle → mid = (left + right) // 2
3.	Compare Target

	•	If nums[mid] == target, return mid

	•	If nums[mid] < target, search the right half (left = mid + 1)

	•	If nums[mid] > target, search the left half (right = mid - 1)

4.	Repeat Until left > right
5.	If target not found, return -1

````python
def bsearch(nums, target):

    nums = sorted(nums)

    # initialize
    left, right = 0, len(nums)-1

    while left <= right:
        mid = (left + right)//2
        print(f"left: {left}, right: {right}")
        print(f"mid index value", mid)
        if nums[mid] == target:
            print("target found")
            return mid
        # if mid >= target
        elif nums[mid] >= target:
            right = mid - 1 # search in left half
            print(f"target is to the left, udpate mid:")
        else:
            left = mid + 1 # search in right half
            print(f"target is to the right, update mid:")

    print("target not found")
    return -1
````

`O(n log n)` denotes an algorithm whose time complexity grows in proportion to the input size `n`  multiplied by the `log` of `n` .

- `Linear Component (n)`: Represents operations that must process each of the  `n`  elements individually.

- `Logarithmic Component (log n)`: Signifies that with each operation, the problem size is effectively halved or reduced by a constant factor.

When combined, `O(n log n)` indicates that the algorithm performs a logarithmic number of operations for each element, leading to a time complexity that scales better than quadratic algorithms (e.g., `O(n²)`) but is less optimal than linear algorithms `(O(n))`.

This notation is commonly associated with efficient sorting algorithms and certain divide-and-conquer strategies.

#### 15 Depth-First Search (DFS)

DFS explores as far as possible along each branch before backtracking, using a stack (explicitly or via recursion).

- with recursion

````python
def dfs(graph, start, visited=None):

    # keep track of every node algorithm visits
    if visited is None:
        visited = set()
        
    visited.add(start)
    print("start", start)
    print("visited", visited)
    
    for neighbor in graph[start]:
        print("neighbor", neighbor)
        if neighbor not in visited:
            dfs(graph, neighbor, visited)
````

- without recursion

````python
def dfs_iterative(graph, start):
    visited = set()
    stack = [start] # Using a Python list as a stack

    while stack:
        vertex = stack.pop() # Pop the last item (LIFO)
        
        if vertex not in visited:
            print(vertex)
            visited.add(vertex)
            
            # Add unvisited neighbors to the stack
            for neighbor in graph[vertex]:
                if neighbor not in visited:
                    stack.append(neighbor)
````

#### 16 Breadth-First Search (BFS)

BFS explores all neighbors at the present depth prior to moving on to nodes at the next depth level, using a queue.

````python
from collections import deque

def bfs(graph, start):
    visited = set()

    # double ended queue initialized with your starting node
    queue = deque([start])
    visited.add(start)
    
    while queue:
        vertex = queue.popleft()
        print("vertex:", vertex)
        
        for neighbor in graph[vertex]:
            print("neighbor:", neighbor)
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
            print("queue:", queue)
````

**19. Topologic sorting problem** 

You have a job dependency system where each Job has dependencies on other jobs. The `execute()` function needs to run all jobs in the graph while respecting their dependencies (a job should only run after all its dependencies have run). This is essentially a topological sorting problem

````python
class Job:
    def __init__(self, deps, name):
        self.deps = deps
        self.name = name

    def run(self):
        pass

    def __repr__(self):
        dep_names = [dep.name for dep in self.deps]
        return f"Job(name='{self.name}', deps={dep_names})"
````

````python
def execute(graph: set[Job]):
    """
    Execute jobs in a dependency graph, ensuring that each job runs only after all
    its dependencies have been executed.
    
    Args:
        graph (set[Job]): A set of Job objects representing the dependency graph
    """
    print("graph", graph)
    
    # Keep track of jobs that have been executed
    executed = set()
    
    # Define a helper function to recursively execute jobs and their dependencies
    def execute_job(job):
        # Skip if job has already been executed
        if job in executed:
            return
        
        # First execute all dependencies
        for dep in job.deps:
            print("Dependency name:", dep.name)
            execute_job(dep)

        print("Executing job:", job.name)

        # Now that all dependencies have been executed, run this job
        job.run()
        
        # Mark this job as executed
        executed.add(job)
    
    # Execute all jobs in the graph
    for job in graph:
        print("Current job:", job.name)
        execute_job(job)
````

````python
# Create an instance of the class
j1 = Job(deps=[], name='job1')
j2 = Job(deps=[j1], name='job2')
j3 = Job(deps=[j1], name='job3')
j4 = Job(deps=[j2, j3], name='job4')

print({j1, j2, j3, j4})

def print_graph(job, level=0):
    indent = "    " * level
    prefix = "|-- " if level > 0 else ""
    print(f"{indent}{prefix}{job.name}")
    for dep in job.deps:
        print_graph(dep, level + 1)

# Visualize starting from the "sink" node (job4)
print("\nDependency Hierarchy:\n")
print_graph(j4)
````

**20. Given the root of a binary tree, return the length of the diameter of the tree.**

The diameter of a binary tree is the length of the longest path between any two nodes in a tree. This path may or may not pass through the root.

The length of a path between two nodes is represented by the number of edges between them.

````python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        
        self.max_diameter = 0

        # If the node is None, return 0 as the height.
        def dfs(node: Optional[TreeNode]) -> int:
            print('node:', node)

            if not node:
                return 0  

            # Recursively compute the height of the left subtree.
            left_height = dfs(node.left)
            right_height = dfs(node.right)

            print('left height', left_height)
            print('right height', right_height)

            current_diameter = left_height + right_height

            # Update the maximum diameter if the current diameter is greater.
            self.max_diameter = max(self.max_diameter, current_diameter)
            print('max diameter', self.max_diameter)

            # Return the height of the current node.
            return 1 + max(left_height, right_height)

        dfs(root)  # Start the DFS traversal from the root.
        return self.max_diameter  # Return the maximum diameter found.
````

**21. Minimum Remove to Make Valid Parentheses**

Given a string s of `'('` , `')'` and lowercase English characters.

Your task is to remove the minimum number of parentheses `( '('` or `')'`, in any positions `)` so that the resulting parentheses string is valid and return any valid string.

Formally, a parentheses string is valid if and only if:

It is the empty string, contains only lowercase characters, or
It can be written as AB (A concatenated with B), where A and B are valid strings, or
It can be written as (A), where A is a valid string.

````python
class Solution:
    def minRemoveToMakeValid(self, s: str) -> str:
            
            stack = []
            remove = set()
            
            for i, char in enumerate(s):
    
                if char == '(':
                    stack.append(i)
                elif char == ')':
                    if stack: 
                        stack.pop() # if matching closing parenthesis found, pop opening (
                    else: # otherwise add to stack
                        remove.add(i) # add unmatched closing parenthesis
            
            # union of all unmatched opening from stack and closing from remove
            remove = remove.union(set(stack))
    
            result = []
            for i, char in enumerate(s):
                if i not in remove:
                    result.append(char)
            
            # list to string
            return ''.join(result)
````

**22. Best time to buy a stock**

You are given an integer array prices where prices[i] is the price of a given stock on the ith day.

On each day, you may decide to buy and/or sell the stock. You can only hold at most one share of the stock at any time. However, you can buy it then immediately sell it on the same day.

Find and return the maximum profit you can achieve.

 
Example 1:

Input: prices = [7,1,5,3,6,4]

Output: 7

Explanation: Buy on day 2 (price = 1) and sell on day 3 (price = 5), profit = 5-1 = 4.

Then buy on day 4 (price = 3) and sell on day 5 (price = 6), profit = 6-3 = 3.

Total profit is 4 + 3 = 7.

````python
class Solution:
    def maxProfit(self, prices) -> int:
        
        """
        In order to book profit, we want to buy low and sell high. 

        Iterate over prices and calculate positive differences or profits
        """

        total_profit = 0

        # iterate starting from second day
        for i in range(1, len(prices)):
            if prices[i] > prices[i-1]:
                total_profit += prices[i] - prices[i-1]
        
        return total_profit
````

**23. In-order traversal**

Inorder traversal is a type of depth-first traversal that follows the sequence: ￼

	1.	Traverse the left subtree
	2.	Visit the root node
	3.	Traverse the right subtree ￼ ￼

This method ensures that nodes are processed in a hierarchical sequence, making it useful for expression trees and BSTs.  ￼

````python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val: int = 0, left: Optional['TreeNode'] = None, right: Optional['TreeNode'] = None):
        self.val = val          # The value of the node
        self.left = left        # Pointer to the left child node
        self.right = right      # Pointer to the right child node

class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        result = []  # List to store the inorder traversal result

        def inorder(node: Optional[TreeNode]):
            if node:
                inorder(node.left)          # Traverse the left subtree
                result.append(node.val)     # Visit the root node
                inorder(node.right)         # Traverse the right subtree

        inorder(root)  # Initiate the traversal starting from the root
        return result  # Return the final traversal result
````

````python
def multiply_matrices(matrix1, matrix2):
    """
    Multiplies two matrices and returns the result.

    Args:
        matrix1: A list of lists representing the first matrix.
        matrix2: A list of lists representing the second matrix.

    Returns:
        A list of lists representing the product of the two matrices, or None if the matrices
        cannot be multiplied.
    """

    # Get the dimensions of the matrices.
    rows1 = len(matrix1)
    # The number of columns in matrix1 is the length of its first row.
    cols1 = len(matrix1[0]) if rows1 > 0 else 0
    
    rows2 = len(matrix2)
    # The number of columns in matrix2 is the length of its first row.
    cols2 = len(matrix2[0]) if rows2 > 0 else 0

    # --------------------------------------------------------------------------------
    # Step 1: Check for valid dimensions.
    # Matrix multiplication is only possible if the number of columns in the first matrix
    # is equal to the number of rows in the second matrix.
    # --------------------------------------------------------------------------------
    if cols1 != rows2:
        print("Error: The number of columns in the first matrix must equal the number of rows in the second matrix.")
        return None

    # --------------------------------------------------------------------------------
    # Step 2: Initialize the result matrix with zeros.
    # The result matrix will have dimensions (rows1 x cols2) - OUTER DIMENSIONS
    # We create a nested list with the correct dimensions and fill it with zeros.
    # --------------------------------------------------------------------------------
    result = [[0 for _ in range(cols2)] for _ in range(rows1)]

    # --------------------------------------------------------------------------------
    # Step 3: Perform the multiplication using three nested loops.
    # The multiplication of two matrices A and B to get matrix C is defined as:
    # C[i][j] = sum(A[i][k] * B[k][j]) for k from 0 to n-1, where n is the number of columns in A (and rows in B).
    #
    # The outer loops iterate through each element of the result matrix.
    # The inner loop performs the dot product calculation for each element.
    # --------------------------------------------------------------------------------
    # Loop over the rows of the first matrix.
    for i in range(rows1):
        # Loop over the columns of the second matrix.
        for j in range(cols2):
            # Loop over the columns of the first matrix (which are the rows of the second matrix).
            for k in range(cols1):
                # Calculate the element at result[i][j] by summing the products
                # of the corresponding elements from the i-th row of matrix1 and the j-th column of matrix2.
                result[i][j] += matrix1[i][k] * matrix2[k][j]

    return result
````

**28. Deserialize a string**

with a `int` or a `list` with nested structure. 

````python
def deserialize(s: str):

    # ----------------------------------------------------------------
    # CASE 1: Simple integer — no brackets means it's just a number
    #
    # If the string doesn't start with '[', it's a flat integer like
    # "42" or "-7". Parse and return it directly — no stack needed.
    #
    # Examples:
    #   "42"   → 42
    #   "-7"   → -7
    # ----------------------------------------------------------------
    if not s.startswith('['):
        return int(s)

    # ----------------------------------------------------------------
    # CASE 2: Nested list — string starts with '['
    #
    # CORE IDEA: Use a stack of lists.
    # Each time we open a '[', we push a new list onto the stack.
    # Each time we close a ']', we pop — the list is complete and
    # already attached to its parent (the list below it on the stack).
    #
    # stack  — each entry is a list currently being built
    #          stack[-1] is always the INNERMOST active list
    #
    # num    — accumulates digit characters to form multi-digit or
    #          negative numbers before converting to int
    #
    # Example: "[1,[2,3]]"
    #   After '[':     stack = [[]]
    #   After '1,':    stack = [[1]]
    #   After '[':     stack = [[1], []]       ← inner list pushed
    #   After '2,':    stack = [[1], [2]]
    #   After '3':     stack = [[1], [2,3]]
    #   After ']':     stack = [[1,[2,3]]]     ← inner list popped & attached
    #   After ']':     stack = [[1,[2,3]]]     ← outer list stays (len==1)
    #   return stack[0] → [1, [2, 3]]
    # ----------------------------------------------------------------
    stack = []
    num = ""

    for i, char in enumerate(s):

        # ------------------------------------------------------------
        # '[' → START OF A NEW LIST
        #
        # Create a fresh empty list for this nesting level.
        #
        # If the stack already has something in it, this new list is
        # NESTED inside the current innermost list — so append it as
        # a child BEFORE pushing. This is what links parent to child:
        #   stack[-1].append(new_list)
        #
        # Then push new_list onto the stack so it becomes the active
        # list that future numbers and sublists attach to.
        #
        # Example: parsing "[1,[2,3]]", at the second '['
        #   stack = [[1]]           ← outer list is active
        #   new_list = []
        #   stack[-1].append([])    → outer list becomes [1, []]
        #   stack.append(new_list)  → stack = [[1,[]], []]
        #                                              ↑ now active
        # ------------------------------------------------------------
        if char == '[':
            new_list = []
            if stack:
                stack[-1].append(new_list)  # attach to parent BEFORE pushing
            stack.append(new_list)          # push to make it the active list

        # ------------------------------------------------------------
        # DIGIT or '-' → BUILDING A NUMBER CHARACTER BY CHARACTER
        #
        # Numbers can be:
        #   - Multi-digit:  "123"  → '1','2','3' accumulated → int("123")
        #   - Negative:     "-42"  → '-','4','2' accumulated → int("-42")
        #
        # We don't convert yet — we wait until we hit ',' or ']'
        # which signals the number is fully formed.
        #
        # Example: char sequence '1','2','3'
        #   num = "" → "1" → "12" → "123"
        # ------------------------------------------------------------
        elif char.isdigit() or char == '-':
            num += char  # accumulate characters, convert later

        # ------------------------------------------------------------
        # ',' or ']' → BOUNDARY: a number or list just ended
        #
        # Both characters signal "whatever number we were building
        # is now complete." We handle the number first, then check
        # if we also need to close a list.
        # ------------------------------------------------------------
        elif char in ',]':

            # --------------------------------------------------------
            # FLUSH NUMBER: if we were building a number, finalize it
            #
            # Convert the accumulated string to int and append it to
            # the current innermost list (stack[-1]).
            # Then reset num = "" for the next number.
            #
            # Why check `if num`?
            # Not every ',' or ']' follows a number. For example:
            #   "[]"    → ']' has no number before it
            #   "[[1]]" → the first ']' closes the inner list, the
            #             second ']' closes the outer — no number before either
            #
            # Example: num="42", stack=[..., [1, 2]]
            #   stack[-1].append(42) → stack=[..., [1, 2, 42]]
            #   num = ""
            # --------------------------------------------------------
            if num:
                stack[-1].append(int(num))
                num = ""

            # --------------------------------------------------------
            # CLOSE LIST: if ']' and stack has more than one list,
            # the current innermost list is complete — pop it.
            #
            # Why `len(stack) > 1`?
            # The outermost list must STAY on the stack so we can
            # return it at the end. We only pop when closing a nested
            # list that has a parent below it.
            #
            # We don't need to "attach" it to the parent here —
            # that was already done when we pushed it (via
            # stack[-1].append(new_list) in the '[' branch).
            # Popping just means we're done filling it.
            #
            # Example: "[1,[2,3]]" at second ']'
            #   stack = [[1,[2,3]], [2,3]]   ← [2,3] is complete
            #   stack.pop()
            #   stack = [[1,[2,3]]]          ← outer list is active again
            # --------------------------------------------------------
            if char == ']' and len(stack) > 1:
                stack.pop()

    # ----------------------------------------------------------------
    # RETURN THE OUTERMOST LIST
    #
    # After the full parse, stack[0] is the top-level list.
    # All nested lists are already attached as children inside it.
    #
    # Example: "[1,[2,[3]]]"
    #   stack[0] = [1, [2, [3]]]
    # ----------------------------------------------------------------
    return stack[0]
````

**29. Streaming Metrics over a Lookback Window**

Build a StreamingMetrics class that ingests ASREvent objects (each with a timestamp, user_id, and num_words) and answers two queries over a sliding time window of the last N seconds:

user_with_biggest_sentence — return the user_id of the event with the most words within the lookback window.
unique_users — return the count of distinct users seen within the lookback window.

Events outside the window (older than now - lookback_seconds) should be excluded from both queries.

````python
from collections import deque

class ASREvent:
    def __init__(self, timestamp, user_id, num_words):
        self.timestamp = timestamp
        self.user_id = user_id
        self.num_words = num_words

class StreamingMetrics:
    def __init__(self, lookback_seconds):
        self.lookback = lookback_seconds
        self.events = deque()

    def record(self, event):
        self.events.append(event)
        self._evict(event.timestamp)

    def _evict(self, now):
        while self.events and self.events[0].timestamp < now - self.lookback:
            self.events.popleft()

    def user_with_biggest_sentence(self, now):
        self._evict(now)
        if not self.events:
            return None
        return max(self.events, key=lambda e: e.num_words).user_id

    def unique_users(self, now):
        self._evict(now)
        return len({e.user_id for e in self.events})
````

````python
metrics = StreamingMetrics(lookback_seconds=10)

metrics.record(ASREvent(timestamp=1, user_id="a", num_words=5))
metrics.record(ASREvent(timestamp=2, user_id="b", num_words=8))
metrics.record(ASREvent(timestamp=3, user_id="a", num_words=3))

assert metrics.user_with_biggest_sentence(now=3) == "b"
assert metrics.unique_users(now=3) == 2

metrics.record(ASREvent(timestamp=15, user_id="c", num_words=2))

assert metrics.user_with_biggest_sentence(now=15) == "c"
assert metrics.unique_users(now=15) == 1

print("ok")
````

**30. Design Underground train system**

An underground railway system is keeping track of customer travel times between different stations. They are using this data to calculate the average time it takes to travel from one station to another.

Implement the UndergroundSystem class:

`void checkIn(int id, string stationName, int t)`

A customer with a card ID equal to id, checks in at the station `stationName` at time `t`.
A customer can only be checked into one place at a time.

`void checkOut(int id, string stationName, int t)`

A customer with a card ID equal to id, checks out from the station `stationName` at time `t`.

`double getAverageTime(string startStation, string endStation)`

Returns the average time it takes to travel from startStation to endStation.

The average time is computed from all the previous traveling times from startStation to endStation that happened directly, meaning a check in at startStation followed by a check out from endStation.

The time it takes to travel from startStation to endStation may be different from the time it takes to travel from endStation to startStation.

There will be at least one customer that has traveled from startStation to endStation before getAverageTime is called.

You may assume all calls to the checkIn and checkOut methods are consistent. If a customer checks in at time t1 then checks out at time t2, then t1 < t2. All events happen in chronological order.

````python
from collections import defaultdict

class UndergroundSystem:

    def __init__(self):
        
        # initialize dict to map cust_id -> (start_station, check_in_time)
        self.check_ins = {}

        # Auto initialize missing keys with defaultdict, no need to check if a key exists
        self.total_time = defaultdict(int) # sum of duration
        self.trip_count = defaultdict(int) # number of trips

    def checkIn(self, id: int, stationName: str, t: int) -> None:
        self.check_ins[id] = (stationName, t)
        print("checkin event:")
        print(self.check_ins)

    def checkOut(self, id: int, stationName: str, t: int) -> None:
        # lookup check_in
        start_station, start_time = self.check_ins.pop(id)
        duration = t - start_time

        route = (start_station, stationName)
        self.total_time[route] += duration
        self.trip_count[route] += 1
        
    def getAverageTime(self, startStation: str, endStation: str) -> float:

        route = (startStation, endStation)
        
        total_time, trip_count = self.total_time[route], self.trip_count[route]
        return total_time / trip_count
````

**31. Design a system that manages the reservation state of n seats that are numbered from 1 to n.**

Implement the SeatManager class:

`SeatManager(int n)` Initializes a SeatManager object that will manage n seats numbered from 1 to n. All seats are initially available.

`int reserve()` Fetches the smallest-numbered unreserved seat, reserves it, and returns its number.

`void unreserve(int seatNumber)` Unreserves the seat with the given seatNumber.

````python
import heapq

class SeatManager:
    def __init__(self, n: int):
        # Min-heap of seats that were reserved and then released.
        # Only contains "holes" below self.next_seat that need to be refilled first.
        self.available = []
        
        # The next seat number that has never been handed out.
        # All seats >= this value are implicitly available.
        self.next_seat = 1
        
        # Upper bound (for clarity / potential validation).
        self.n = n

    def reserve(self) -> int:
        # Prefer reusing the smallest released seat if one exists,
        # otherwise hand out the next fresh seat.
        print("Available list:", self.available)
        if self.available:
            # Find min available seat
            min_avail = heapq.heappop(self.available)
            print("Min Available:", min_avail)
            return min_avail  # O(log k), k = heap size
        
        seat = self.next_seat
        print("Next available seat:", seat)
        self.next_seat += 1
        return seat

    def unreserve(self, seatNumber: int) -> None:
        # Push the released seat back into the heap so it can be reused by the next reserve() call.
        heapq.heappush(self.available, seatNumber)  # O(log k)

        return self.available
````

- Finds the latest price of the stock based on the current records. The latest price is the price at the latest timestamp recorded.

- Finds the minimum price the stock has been based on the current records.

**36. LRU Cache**

Design a data structure that follows the constraints of a `Least Recently Used (LRU)` cache.

Implement the `LRUCache class`:

`LRUCache(int capacity)` Initialize the LRU cache with positive size capacity.

`int get(int key)` Return the value of the key if the key exists, otherwise return -1.

`void put(int key, int value)` Update the value of the key if the key exists. Otherwise, add the key-value pair to the cache. If the number of keys exceeds the capacity from this operation, evict the least recently used key.

Track recency ordering as accesses happen, so we know which key is "least recently used" at eviction time.

The functions get and put must each run in `O(1)` average time complexity.

````python
from collections import OrderedDict

class LRUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = OrderedDict()

    def get(self, key: int) -> int:
        """
        look up by key, return value, and mark this key as most recently used.
        """ 
        if key not in self.cache:
            return -1
            
        """
        Move it to the end of the OrderedDict to mark it as the most recently used. `move_to_end` is `O(1)` because OrderedDict maintains a doubly-linked list
        """
        self.cache.move_to_end(key)
        return self.cache[key]

    def put(self, key: int, value: int) -> None:
        """
        insert or update. If at capacity and inserting new, evict the least recently used key first.
        """
        # if key exists, promote it to MRU position
        if key in self.cache:
            self.cache.move_to_end(key)

        self.cache[key] = value

        # if exceeded capacity, evict the oldest entry
        if len(self.cache) > self.capacity:
            oldest_key = next(iter(self.cache))
            print(f"capacity exceeded, evicted: {self.cache[oldest_key]}")
            self.cache.popitem(last=False)
````

````python
def test_lru_cache():
    cache = LRUCache(2)
    
    cache.put(1, 100)
    cache.put(2, 200)
    assert cache.get(1) == 100      # 1 is now MRU, 2 is LRU
    
    cache.put(3, 300)               # evicts 2 (least recently used)
    assert cache.get(2) == -1       # 2 is gone
    assert cache.get(3) == 300      # 3 is present
    
    cache.put(1, 999)               # update existing key, promotes 1 to MRU
    cache.put(4, 400)               # evicts 3 (now LRU after 1 was promoted)
    assert cache.get(3) == -1       # 3 evicted
    assert cache.get(1) == 999      # 1 still here with updated value
    assert cache.get(4) == 400


if __name__ == "__main__":
    test_lru_cache()
    print("Passed.")
````

**37. Building a Production-Style In-Memory Key-Value Store**

Understanding what we're building. 

We're going to construct a key-value store that grows in capability across six levels. Each level adds a real-world concern that turns a toy data structure into something resembling production infrastructure `(Redis, Memcached, DynamoDB-style stores)`. 

The progression matters: each level introduces a problem that the previous design can't handle cleanly, and we solve it by reaching for the right data structure.

Here's what we're solving at each level:

**Level 1:** Basic operations. A simple dictionary-backed store. The starting point — most things should work, most code should be obvious.

**Level 2:** TTL (time-to-live). Real caches need automatic expiration. A session token expires in 30 minutes; a rate-limit counter resets every minute. We need a way for entries to disappear on their own, without external cleanup.

**Level 3:** Prefix scans. Need to find all keys starting with `"user:123:"` or `"session:"`. A plain dict can't do this efficiently — you'd scan every key, which is `O(n)` per query. We need an ordering structure.

**Level 4:** Backup and restore. Production stores need point-in-time recovery. If a bug corrupts data at 3pm, you want to restore to 2:30pm. This is the same "versioned history" pattern from SnapshotArray and TimeMap, but applied to the whole store.

**Level 5:** Thread safety. Concurrent writers can corrupt shared state. Two threads calling set(key, value) at the same moment can interleave their reads and writes in ways that break invariants. We need locking — and we need to be careful which operations need it.

**Level 6:** Pagination. A scan that returns 10 million matching keys at once will OOM the client. Real APIs return pages with cursors. We need to support resumable iteration.

````python
class KVStore:

    # initialize a standard dict 
    def __init__(self):
        self._store = {}

    # set value to the given key
    def set(self, key: str, value):
        self._store[key] = value

    # retrieve value 
    def get(self, key: str):
        val = self._store[key]
        return val
````

**39. Checkered Flood Fill (BFS Distance-Parity Coloring)**

The problem is a custom grid-traversal challenge known as the Checkered Flood Fill.

Given a `2D` grid (matrix) of cells, a starting coordinate (sr, sc), and a palette of colors, the goal is to traverse all reachable, non-blocked cells and fill them with an alternating, two-tone pattern—resembling a chessboard or checkerboard.

Key Rules of the ProblemThe Matrix ($m$): A 2D grid where each cell contains a specific state or color defined by an Enum (such as `BLACK`, `WHITE`, or `RED`).

The Obstacles: Cells marked as Color. BLACK act as solid walls. The filling algorithm cannot pass through them, color them, or see past them. The Starting Point: The filling begins at a specified seed coordinate `(sr, sc)`. If this cell is already a wall (BLACK), the algorithm terminates immediately. 

The Checkered Effect: Instead of traditional flood fill algorithms that paint an entire connected component a single solid color, this problem requires the output to alternate colors based on how far away a cell is from the starting point. The Mathematical Core: Distance Parity The problem relies on shortest-path distance parity (whether the number of steps from the start is even or odd) to determine the color of each cell:$$\text{Cell Color} = \begin{cases} \text{Color 1 (e.g., RED)}, & \text{if } \text{distance} \text{ is even} \\ \text{Color 2 (e.g., WHITE)}, & \text{if } \text{distance} \text{ is odd} \end{cases}$$. 

Because it requires alternating patterns based on accurate step counts, a layer-by-layer traversal strategy like Breadth-First Search (BFS) is uniquely suited to solve it, ensuring that the shortest distance to every cell is locked in before it receives its final color.

````python
from collections import deque
from enum import Enum

class Color(Enum):
    BLACK = "B"
    WHITE = "W"
    RED = "R"


def fill(m, sr=0, sc=0):
    if not m or not m[0] or m[sr][sc] is Color.BLACK:
        return m
    h, w = len(m), len(m[0])
    palette = (Color.RED, Color.WHITE)
    seen = [[False] * w for _ in range(h)]
    seen[sr][sc] = True
    q = deque([(sr, sc, 0)])
    while q:
        r, c, d = q.popleft()
        m[r][c] = palette[d & 1]
        for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1)):
            nr, nc = r + dr, c + dc
            if 0 <= nr < h and 0 <= nc < w and not seen[nr][nc] and m[nr][nc] is not Color.BLACK:
                seen[nr][nc] = True
                q.append((nr, nc, d + 1))
    return m

def board(rows):
    return [[Color(ch) for ch in row] for row in rows]

def render(m):
    return "\n".join(" ".join(c.value for c in row) for row in m)
````

**40. Autocomplete Name Suggestion**

1. Visualizing the Trie (Prefix Tree)
Think of a Trie as a roadmap of characters. Every node represents a letter, and paths from the root down to marked nodes form complete names. Let's see how the names `"Abraham"`, `"Abby"`, and `"April"` look when stored inside a Trie.

2. How Autocomplete Works Using This Diagram:

If a user types `"abr"`, the system starts at the (Root). It follows the path: a $\rightarrow$ b $\rightarrow$ r.Once it reaches the `r` node, it explores all possible branches beneath it (using a Depth-First Search) to gather completed names. In this case, it follows the only remaining path down to `m*` and suggests "Abraham". If the user had typed `"ab"`, it would look down both branches under the second `b` and suggest both `"Abraham"` and `"Abby"`.


3. Step-by-Step Algorithm
    
To implement this, our program needs to perform three primary actions:

- Insert: Break each name down into characters and insert them node-by-node into the tree. Store the full, original name at the final leaf node for easy retrieval.

- Find Prefix: Traverse the tree matching the characters of the user's input prefix `(e.g., "abr")`. If we hit a dead end, it means no names match.

- Gather Suggestions: Once we find the node corresponding to the end of the prefix, traverse all of its child nodes recursively to harvest every full name down the line.

```
(Root)
                 |
                 a
               /   \
              b     p
             / \     \
            r   b     r
           /     \     \
          a       y* i
          |             |
          h             l*
          |
          a
          |
          m*
          
   (* indicates the point where a complete name ends)
```

````python
class TrieNode:
    def __init__(self):
        # Maps a character to its corresponding child TrieNode
        self.children = {}
        # Stores the original name if this node represents the end of a valid word
        self.suggested_name = None

class AutocompleteSystem:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, name: str):
        """Inserts a name into the Trie (case-insensitive for matching)"""

        # set the start point
        node = self.root
        
        # Convert to lowercase to ensure case-insensitive autocomplete matching
        for char in name.lower():
            # check if char in current node's children
            if char not in node.children:
                node.children[char] = TrieNode() # create a new branch in Trie when char doesn't exist
            # move the pointer down the tree into the child node you either checked or created
            node = node.children[char]
            print("current char", char)
            
        # Store the original correctly-cased name at the final node
        self.suggested_name = name
        print("names:", self.suggested_name)


    def _dfs_gather(self, node: TrieNode, results: list):
        """Helper method to collect all names down the tree using Depth-First Search"""
        if node.suggested_name:
            results.append(node.suggested_name)
        
        # Recursively visit all child branches alphabetically
        for char in sorted(node.children.keys()):
            self._dfs_gather(node.children[char], results)

    def autocomplete(self, prefix: str) -> list:
        """Returns a list of all names starting with the given prefix"""
        node = self.root
        
        # 1. Navigate to the end of the prefix
        for char in prefix.lower():
            if char not in node.children:
                return []  # Prefix not found, return empty suggestions
            node = node.children[char]
        
        # 2. Collect all names branching off from this prefix node
        suggestions = []
        self._dfs_gather(node, suggestions)
        return suggestions
````
