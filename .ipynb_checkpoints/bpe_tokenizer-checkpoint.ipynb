# Byte Pair Encoding (BPE) Tokenizer — From Scratch

**Core idea:** BPE starts with individual characters as tokens, then iteratively merges the most frequent adjacent pair into a new token. Repeat until you hit your target vocabulary size.

This is the same algorithm used by GPT-2, GPT-4, and most modern LLMs to build their vocabularies.

### Steps
1. Split corpus into words → split words into characters
2. Count all adjacent token pairs (bigrams)
3. Merge the most frequent pair into a new token
4. Repeat steps 2–3 for `num_merges` iterations
5. Use the learned merge rules to tokenize new text

```python
import re
from collections import defaultdict
```

## Step 1 — Build the initial vocabulary

```python
def get_vocab(corpus: str) -> dict:
    """
    Convert raw text into a word-frequency dictionary.
    Each word is stored as space-separated characters + '</w>' end-of-word marker.
    The '</w>' marker lets the model distinguish 'low' inside 'lower'
    from 'low' as a standalone word.
    """
    
    vocab = defaultdict(int)
    
    for word in corpus.split(): # split corpus on whitespace, convert raw text to list
        # join chars with spaces so each char is a separate token,
        # append '</w>' to mark where the word ends
        token_word = " ".join(list(word)) + " </w>" # concatenate items in an iterable
        vocab[token_word] += 1 # count how often this word appears
    return dict(vocab)
```

```python
# --- quick test ---
corpus = "low lower newest widest low low lower newest widest newest low lower newest widest widest widest"
vocab = get_vocab(corpus)
print("vocab:", vocab)

print("Initial vocabulary (word → frequency):")
for word, freq in sorted(vocab.items(), key=lambda x: -x[1]):
    print(f"  {word!r:35s}  freq={freq}")
```

## Step 2 — Count adjacent token pairs (bigrams)

```python
def get_pair_counts(vocab: dict) -> dict:
    """
    Scan every word in the vocabulary and tally how often each
    consecutive pair of tokens appears, weighted by word frequency.
    e.g. 'l o w </w>' with count 3 contributes ('l','o'):3, ('o','w'):3 ...
    """
    
    pairs = defaultdict(int)
    
    for word, freq in vocab.items():
        symbols = word.split()                    # ['l', 'o', 'w', '</w>']
        
        for i in range(len(symbols) - 1):         # slide a window of size 2
            pairs[(symbols[i], symbols[i + 1])] += freq   # weight by word freq
    return dict(pairs)
```

```python
# --- quick test ---
pairs = get_pair_counts(vocab)
print("pairs:", pairs)

print("Top 10 most frequent pairs:")
for pair, count in sorted(pairs.items(), key=lambda x: -x[1])[:10]:
    print(f"  {str(pair):25s}  count={count}")
```

## Step 3 — Merge the best pair everywhere

```python
def merge_vocab(pair: tuple, vocab: dict) -> dict:
    """
    Given the best pair (e.g. ('e', 's')), replace every occurrence of
    'e s' (space-separated) with 'es' (merged) in every vocabulary entry.
    We use a regex to only match the exact pair (not substrings of longer tokens).
    """
    
    new_vocab = {}
    
    # build a regex that matches the pair surrounded by word boundaries
    # re.escape handles special chars like '<' and '>'
    bigram = re.escape(" ".join(pair))            # e.g. r"e s"
    pattern = re.compile(r"(?<!\S)" + bigram + r"(?!\S)")  # exact token match

    #print("bigram", bigram)
    #print("pattern", pattern)

    for word in vocab:
        # replace 'e s' → 'es' everywhere this pair appears
        new_word = pattern.sub("".join(pair), word)
        new_vocab[new_word] = vocab[word]
    return new_vocab
```

```python
# --- quick test: merge the top pair ---
best = max(pairs, key=pairs.get)
print(f"Best pair to merge: {best}  (count={pairs[best]})")
vocab_after = merge_vocab(best, vocab)

print("\nVocab after one merge:")
for word, freq in sorted(vocab_after.items(), key=lambda x: -x[1]):
    print(f"  {word!r:35s}  freq={freq}")
```

## Step 4 — Train BPE (run all merge steps)

```python
def train_bpe(corpus: str, num_merges: int) -> tuple:
    """
    Train BPE on `corpus` for `num_merges` merge steps.
    Returns:
      - final vocabulary  (word → frequency)
      - merges list       (ordered list of merged pairs — this IS the model)
    """
    vocab = get_vocab(corpus)                     # start: each char is its own token
    print("vocab:", vocab)
    merges = []                                   # will store merge rules in order

    for i in range(num_merges):
        pairs = get_pair_counts(vocab)            # count all bigrams
        if not pairs:
            break                                 # corpus fully merged, nothing left
        print("bigrams", pairs)
        
        # pick the pair with the highest frequency
        best_pair = max(pairs, key=pairs.get)
        print("best pair", best_pair)
        
        vocab = merge_vocab(best_pair, vocab)     # apply the merge to the vocab
        merges.append(best_pair)                  # record the merge rule
        
        print(f"Merge {i+1:3d}: {best_pair[0]!r:10s} + {best_pair[1]!r:10s}"
              f" → {(best_pair[0]+best_pair[1])!r:12s}  (freq={pairs[best_pair]})")

        print("merges:", merges)
        print("\n")

    return vocab, merges
```

```python
print("=== Training BPE (15 merges) ===")
final_vocab, merges = train_bpe(corpus, num_merges=15)
```

## Step 5 — Inspect final vocabulary

```python
print("\n=== Final Vocabulary ===")
for word, freq in sorted(final_vocab.items(), key=lambda x: -x[1]):
    tokens = word.split()
    print(f"  {word!r:30s}  → tokens={tokens}  freq={freq}")

print(f"\nTotal merge rules learned: {len(merges)}")
print(f"Merge rules: {merges}")
```

## Step 6 — Tokenize new text using learned merge rules

```python
def tokenize(text: str, merges: list) -> list:
    """
    Apply the learned merge rules (in order) to new text.
    This is inference — we replay the exact same merges from training.
    """
    words = text.split()
    tokenized = []

    for word in words:
        chars = list(word) + ["</w>"]             # ['h','e','l','l','o','</w>']

        # apply every merge rule in the order they were learned
        for (left, right) in merges:
            i = 0
            new_chars = []
            while i < len(chars):
                # if current and next token match the pair, fuse them
                if i < len(chars) - 1 and chars[i] == left and chars[i+1] == right:
                    new_chars.append(left + right) # merge: 'h'+'e' → 'he'
                    i += 2                         # skip both consumed tokens
                else:
                    new_chars.append(chars[i])     # no match, keep as-is
                    i += 1
            chars = new_chars                      # update for next merge rule

        tokenized.extend(chars)                    # flatten into output token list

    return tokenized


print("=== Tokenizing new text ===")
test_sentences = [
    "low newest widest",
    "lower widest",
    "newest low",
]

for sentence in test_sentences:
    tokens = tokenize(sentence, merges)
    print(f"  Input : {sentence!r}")
    print(f"  Tokens: {tokens}")
    print()
```

## Experiment — Try your own corpus and merge count

```python
# ---- tweak these and re-run ----
my_corpus = """
the cat sat on the mat the cat ate the rat
the rat ran and the cat ran after the rat
"""
num_merges = 20
# --------------------------------

my_vocab, my_merges = train_bpe(my_corpus, num_merges=num_merges)

print("\n=== Tokenize a new sentence ===")
result = tokenize("the cat sat", my_merges)
print(result)
```
